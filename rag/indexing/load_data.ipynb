> load data from a data source and standardize into a langchain doc

> data is stored as a list of document objects

#### PDF Loader

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
pdf_loader = PyPDFLoader('../../data/Introduction_to_Data_and_Data_Science.pdf')
```

```python
docs = pdf_loader.load() # entire pdf is converted to list of documents with each page as one document
```

```python
print('Number of docs:', len(docs))
print(type(docs[0]))

docs[2].metadata
```

#### Docx2text Loader

```python
from langchain_community.document_loaders import Docx2txtLoader
```

```python
docx_loader = Docx2txtLoader('../../data/Introduction_to_Data_and_Data_Science.docx')
```

```python
docs = docx_loader.load()
```

```python
print(len(docs))
docs[0].metadata
```

#### Preprocess

> different documents require different preprocessing

> in our case, both the pdf & docx isn't well-formatted and have too many `'\n'` chars

> remove newline chars `'\n'` from the page content of each doc to decrease token consumption

```python
docs[0].page_content[:100]
```

```python
import re

for doc in docs:
    doc.page_content = re.sub('\n', ' ', doc.page_content)
```
