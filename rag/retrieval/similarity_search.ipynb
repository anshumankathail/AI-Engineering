#### Load Vector DB

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
embedding_fn = HuggingFaceEmbeddings(model_name = "BAAI/bge-base-en-v1.5")
```

```python
vectorstore = Chroma(embedding_function = embedding_fn, persist_directory = '../../vectordb/intro_to_data_science')
```

#### Retrieval

```python
question = "What are the tools and programming languages used by data scientists?"
```

```python
retrieved_docs = vectorstore.similarity_search(query = question, k = 4)
```

```python
for doc in retrieved_docs:
    print(doc.metadata['Topic Title'], '\n', doc.page_content, end = '\n\n')
```

> drawback: usually returns highly similar or redundant chunks from the vector db

> i.e. there is low diversity among the retrieved chunks
