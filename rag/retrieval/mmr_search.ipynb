#### Load Vector DB

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
embedding_fn = HuggingFaceEmbeddings(model_name = "BAAI/bge-base-en-v1.5")
```

```python
vectorstore = Chroma(embedding_function = embedding_fn, persist_directory = '../../vectordb/intro_to_data_science')
```

#### Maximal Marginal Relevance Search

marginal relevance = λ . relevance(query, document) - (1-λ) . Max(relevance(document, retrived_documents))

- λ ∈ [0, 1]

- λ closer to 0 means the focus is more on diversity and λ closer to 1 means focus is more on relevance

```python
question = "What software do data scientists use? "
```

```python
retrieved_docs = vectorstore.max_marginal_relevance_search(query = question, k = 3, lambda_mult = 0.5)
```

```python
for doc in retrieved_docs:
    print(doc.metadata['Topic Title'], '\n', doc.page_content, '\n\n')
```
