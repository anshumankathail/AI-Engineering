```python
import os
import whisper
import pandas as pd
```

```python
directory_path = '../data/recordings'
```

```python
model = whisper.load_model('base')
```

#### Transcribe multiple files

```python
def transcribe_directory(directory_path):
    transcriptions = []

    for file_name in os.listdir(directory_path):
        if file_name.endswith('.wav'):
            file_path = os.path.join(directory_path, file_name)
            result = model.transcribe(file_path)

            transcriptions.append({'File Name': file_name, 'Transcription': result["text"]})

    return transcriptions
```

```python
transcriptions = transcribe_directory(directory_path)
```

#### Save transcriptions as csv

Saving as a csv allows for future analysis like sentiment analysis, keyword extraction or training an ML model

```python
file_path = '../data/transcriptions.csv'
```

```python
df = pd.DataFrame(transcriptions)
```

```python
df.rename(columns={'File Name': 'Track'}, inplace=True)
```
