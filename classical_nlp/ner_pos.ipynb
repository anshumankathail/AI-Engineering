```python
import spacy
import pandas as pd
```

```python
nlp = spacy.load('en_core_web_sm')
```

##### import data

```python
df = pd.read_csv('../data/bbc_news.csv')
df.head(2)
```

```python
text = ' '.join(df['title'] + '. ' + df['description']) # convert pandas series inside the parantheses to a single long string
print(text[:200])
```

```python
doc = nlp(str(text))
```

> spacy should usually be fed raw text

> spacy tokenizes text, gives POS and NER tags, ...

##### Named Entitiy Recognition

```python
from spacy import displacy

displacy.render(doc[:100], style = 'ent')
```

```python
ner_df = pd.DataFrame(columns = ['token', 'tag'])

for word in doc.ents:
    ner_df.loc[len(ner_df)] = [word.text, word.label_]

ner_df.head()
```

```python
ner_df_grouped = ner_df.value_counts(['token', 'tag']).sort_values(ascending = False).reset_index(name = 'count') # group by token and tag

ner_df_grouped.head(10)
```

```python
ner_df_grouped[ner_df_grouped['tag'] == 'GPE'].head(10)
```

##### Part of Speech tagging

```python
pos_df = pd.DataFrame(columns = ['token', 'tag'])

for token in doc:
    pos_df.loc[len(pos_df)] = [token.text, token.pos_]

pos_df.head()
```

```python
pos_df_grouped = pos_df.value_counts(['token', 'tag']).sort_values(ascending = False).reset_index(name = 'count') # group by token and tag

pos_df_grouped.head()
```

```python
pos_df_grouped[pos_df_grouped['tag'] == 'NOUN'].head(10)
```
