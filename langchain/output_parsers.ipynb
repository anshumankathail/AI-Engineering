```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
```

##### Chat Model

```python
chat = ChatGoogleGenerativeAI(
    model = "gemini-3.1-flash-lite-preview",
    temperature = 0,
    max_output_tokens = 50,
    seed = 0
)
```

#### String Output Parser

> StrOutputParser ensures a str output without manual extraction from structured AI Message fields

```python
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
hum_msg = HumanMessage("Tell me an interesting fact that most people don't know.")
```

```python
response = chat.invoke([hum_msg])
response.text
```

```python
str_out_parser = StrOutputParser()
```

```python
response_str = str_out_parser.invoke(response)
response_str
```

#### Comma Separated List Output Parser

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
hum_msg = HumanMessage(f'''Which species have gone extinct recently?

{CommaSeparatedListOutputParser().get_format_instructions()}
''')
```

```python
print(hum_msg.content)
```

```python
response = chat.invoke([hum_msg])
response.text
```

```python
list_out_parser = CommaSeparatedListOutputParser()
```

```python
response_list = list_out_parser.invoke(response)
response_list
```
