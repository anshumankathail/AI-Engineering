```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
load_dotenv
```

##### initialize

```python
chat = ChatGoogleGenerativeAI(
    model = "gemini-3.1-flash-lite-preview",
    temperature = 0,
    max_output_tokens = 50,
    seed = 0
)
```

```python
chat_template = ChatPromptTemplate.from_messages([(
    'human',
    'How to train a {species} of {breed}.'
)])
```

```python
chain = chat_template | chat
```

#### invoke

```python
%%time
response = chain.invoke({'species': 'dragon', 'breed': 'night fury'})
```

```python
%%time
chain.invoke({'species': 'griffin', 'breed': 'dawn wings'})
```

#### batch invoke

```python
%%time
chain.batch([
    {'species': 'dragon', 'breed': 'night fury'},
    {'species': 'griffin', 'breed': 'dawn wings'}
])
```

> the combined wall time of individual chain invoke methods is more than that of batch method

> hence, batch method is faster for multiple inputs as compared to sequential invoking

#### stream invoke

```python
response = chain.stream({'species': 'dragon', 'breed': 'night fury'})
response
```

> generator objects are one-time iteratable objects

```python
import time

for chunk in response:
    print(chunk.text, end = '')
    time.sleep(0.25)
```
