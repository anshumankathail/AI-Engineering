```python
from langchain_core.runnables import RunnableLambda
```

#### Runnable Lambda

> allows us to convert ordinary functions into runnables

> we can then use the methods such as invoke, batch, stream on them

> predefined functions can be passed as an argument to RunnableLambda

> but the function must accept a single input argument (can be a list, dict, tuple, etc.)

##### Runnable 1

```python
runnable_sum = RunnableLambda(lambda x: sum(x))
```

```python
runnable_sum.invoke([1, 2, 3])
```

##### Runnable 2

```python
runnable_square = RunnableLambda(lambda x: x**2)
```

```python
runnable_square.invoke(6)
```

##### Chain

```python
chain1 = runnable_sum | runnable_square
```

```python
chain1.invoke([1, 2, 3])
```

##### Visualize Chain

```python
chain1.get_graph().print_ascii()
```

#### Chain Decorator

@chain: decorator

> a decorator is a function that modifies the behaviour of another function or class

```python
from langchain_core.runnables import chain
```

> here, @chain wraps the ordinary function in RunnableLambda with the same name

```python
@chain
def runnable_sum(x):
    return sum(x)

@chain
def runnable_square(x):
    return x**2
```

```python
type(runnable_sum)
```

```python
chain2 = runnable_sum | runnable_square
```

```python
chain2.invoke([1, 2, 3])
```

##### Visualize Chain

```python
chain2.get_graph().print_ascii()
```
